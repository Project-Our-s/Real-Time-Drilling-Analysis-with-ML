```python
#univariate analysis for weight on bit 
def univariate_analysis(column):

    plt.figure(figsize=(18,5))

    # Histogram
    plt.subplot(1,3,1)
    sns.histplot(df[column], kde=True)
    plt.title(f'Histogram of {column}')

    # Density Plot
    plt.subplot(1,3,2)
    sns.kdeplot(df[column], fill=True)
    plt.title(f'Density Plot of {column}')

    # Boxplot
    plt.subplot(1,3,3)
    sns.boxplot(x=df[column])
    plt.title(f'Boxplot of {column}')

    plt.tight_layout()
    plt.show()

    print(df[column].describe())
```

```python
univariate_analysis("weight_on_bit")
```

The Weight on Bit (WOB) variable exhibits a multimodal and slightly right-skewed distribution, indicating that different drilling conditions required different WOB settings. The boxplot confirms that the preprocessing stage successfully minimized extreme outliers. The interquartile range (2.25–171.88 klbf) represents the typical operating range, while the high standard deviation reflects significant variability in drilling operations.

```python
#Bivariate Analysis
def bivariate_analysis(x, y):

    plt.figure(figsize=(8,6))

    sns.regplot(
        data=df,
        x=x,
        y=y,
        scatter_kws={"alpha":0.2, "s":8},
        line_kws={"color":"red"}
    )

    plt.title(f"{x} vs {y}")
    plt.xlabel(x)
    plt.ylabel(y)

    plt.show()

    print(df[[x,y]].corr())
```

```python
#Pump pressure vs Flow in 
bivariate_analysis("pump_pressure","flow_in")
```

The scatter plot indicates a moderate positive relationship between pump pressure and flow rate, as the regression line slopes upward. The correlation coefficient (r = 0.598) suggests that higher flow rates are generally associated with higher pump pressures. However, the wide spread of data points indicates that additional drilling parameters also influence pump pressure. Overall, the relationship is positive but not perfectly linear.

```python
#Top drive rpm vs ROP depth hour
bivariate_analysis("top_drive_rpm","rop_depth_hour")
```

The rop_depth_hour variable has zero variance, making it impossible to evaluate the relationship between Top Drive RPM and Rate of Penetration (ROP). Consequently, this analysis was excluded.

```python
#hookload vs block position
bivariate_analysis("hookload","block_position")
```

The correlation coefficient (r = -0.028) indicates an extremely weak negative relationship, suggesting that changes in hookload do not have a meaningful linear association with block position. This implies that block movement is influenced by several operational factors beyond hookload alone.

# Engineering Insights

## Key Findings

### Mechanical Insights
- Weight on Bit (WOB) demonstrates a positive relationship with Downhole Torque, indicating that increased bit loading generally requires greater rotational torque.
- Hookload increases with Weight on Bit, suggesting effective transfer of drilling load to the drill bit.
- Top Drive RPM remains relatively stable throughout drilling operations, while Downhole Torque exhibits moderate fluctuations corresponding to changing drilling conditions.

### Hydraulic Insights
- Pump Pressure generally increases with Flow In, indicating a positive hydraulic relationship during drilling.
- Differential Pressure remains stable for most of the operation, with occasional spikes that may correspond to operational adjustments or changing formation characteristics.
- Return Flow remains relatively consistent, suggesting efficient mud circulation and wellbore cleaning.

### Time-Series Insights
- Time-series analysis reveals stable drilling operations with periodic fluctuations in pressure, torque, hookload, and flow parameters.
- No prolonged abnormal operating conditions are observed after preprocessing.

### Correlation Insights
- Correlation analysis identifies meaningful relationships among several drilling parameters, particularly between hydraulic and mechanical variables.
- These relationships provide useful information for feature selection in subsequent machine learning models.

### Operational Insights
- Rig Mode analysis indicates that drilling operations dominate the recorded activities.
- Drill Mode analysis highlights the most frequently occurring drilling states and operational behavior.

### Data Quality
- Outlier visualization confirms that preprocessing successfully reduced extreme observations while preserving the overall data distribution.
- Missing values and invalid entries were appropriately handled during the preprocessing stage.

### Machine Learning Relevance
The exploratory data analysis identifies the most influential drilling parameters and their relationships, providing a strong foundation for predictive modeling, anomaly detection, and drilling performance optimization in the subsequent machine learning notebooks.
